```python
# Filter the dataframe for non-null values of the feature 'Chl-a'
filtered_df = df[df['DOC'].notnull()]

# Group by 'Cluster' and count the number of rows in each cluster
cluster_counts = filtered_df.groupby('Cluster').size()

# Sort the counts in descending order and select the top 5 clusters
top_5_clusters = cluster_counts.sort_values(ascending=False).head(15)
```
