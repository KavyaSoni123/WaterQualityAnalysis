```python
import pandas as pd
```

```python
df1 = pd.read_csv('data/Gemstat_Data_merged_sattelite/Lake Laguna del Sauce(678)/Lake Laguna del Sauce(678)_1_days.csv')
df2 = pd.read_csv('data/Gemstat_Data_merged_sattelite/Lake_erie/Lake_erie_1_days.csv')
df3 = pd.read_csv('data/Gemstat_Data_merged_sattelite/lake_lago_de_paitz(393)/lake_lago_de_paitz(393)_1_days.csv')
```

```python
# Your selected columns
selected_columns = ["TURB", "B11", "B12", "B2", "B3", "B4", "B5", "B6", "B7", "B8", "B8A"]
selected_columns2 = ["Turbidity", "B11", "B12", "B2", "B3", "B4", "B5", "B6", "B7", "B8", "B8A"]

# Extract the subsets
df_selected = df1[selected_columns]
df_selected3 = df3[selected_columns]
df_selected2 = df2[selected_columns2]

# Rename "Turbidity" in df2 to "TURB" for consistency
df_selected2 = df_selected2.rename(columns={"Turbidity": "TURB"})

# Concatenate the dataframes vertically
df_merged = pd.concat([df_selected, df_selected2, df_selected3], axis=0, ignore_index=True)

```

```python
df_merged = df_merged.dropna(subset=["TURB"])
```
